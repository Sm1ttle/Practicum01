```python
from pathlib import Path

from dotenv import load_dotenv
import os
import sys

from openai import AzureOpenAI
```

```python
# Jupyter/Cursor kernels often have a cwd or call stack that is *not* this folder.
# Load .env from the notebook directory so os.getenv() is not None.
_notebook_file = globals().get("__vsc_ipynb_file__")
_notebook_dir = Path(_notebook_file).resolve().parent if _notebook_file else Path.cwd()
_env_path = _notebook_dir / ".env"

loaded = load_dotenv(_env_path, override=True)
print(f"Using {_env_path}")
print(f"File exists: {_env_path.exists()}  |  variables loaded: {loaded}")

def env(name: str):
    value = os.getenv(name)
    if value is None:
        return None
    value = value.strip()
    return value or None

# APIM settings from .env. Endpoint must be the host only, e.g.
# https://apim-azr-ue2-bgpt-prd-ucin.azure-api.net  (no /openai/deployments)
api_key = env("AZURE_APIM_OPENAI_SUBSCRIPTION_KEY")
api_version = env("AZURE_APIM_OPENAI_API_VERSION")
endpoint = env("AZURE_APIM_OPENAI_ENDPOINT")
deployment = env("AZURE_APIM_OPENAI_DEPLOYMENT")

print(
    "Resolved:",
    {
        "api_key": "set" if api_key else "None",
        "api_version": api_version or "None",
        "endpoint": endpoint or "None",
        "deployment": deployment or "None",
    },
)
```

```python
if not all([api_key, api_version, endpoint, deployment]):
    sys.exit(
        "Missing Azure APIM settings. Set AZURE_APIM_OPENAI_SUBSCRIPTION_KEY, "
        "AZURE_APIM_OPENAI_API_VERSION, AZURE_APIM_OPENAI_ENDPOINT, and "
        "AZURE_APIM_OPENAI_DEPLOYMENT in your .env file."
    )

print(f"Azure APIM key exists and begins {api_key[:8]}")
print(f"Deployment: {deployment}")
```
