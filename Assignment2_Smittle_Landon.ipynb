## Question 3 (20 pts) — Lists and Dicts
Given the tool-call log entries below (a list of dicts), using `for` loop:
1. Build a list called `slow_calls` containing `"{name}: {duration}ms"` strings, but only for entries where `duration` is greater than 100.
2. Build a dictionary called `duration_by_tool` that maps each tool's `name` to its `duration`.

Print both `slow_calls` and `duration_by_tool`.

```python
logs = [
    {"name": "search", "duration": 340},
    {"name": "calculator", "duration": 45},
    {"name": "browser", "duration": 210},
    {"name": "vector_db", "duration": 88},
]

logs = [
    {"name": "search", "duration": 340},
    {"name": "calculator", "duration": 45},
    {"name": "browser", "duration": 210},
    {"name": "vector_db", "duration": 88},
]

slow_calls = []
duration_by_tool = {}

for entry in logs:
    if entry["duration"] > 100:
        slow_calls.append(f"{entry['name']}: {entry['duration']}ms")
    duration_by_tool[entry["name"]] = entry["duration"]

print("slow_calls:", slow_calls)
print("duration_by_tool:", duration_by_tool)

slow_calls = None
print(slow_calls)
```

## Question 4 (20 pts) — Control Flow
Write a function `classify_latency(ms)` that returns:
- `"fast"` if `ms < 150`
- `"moderate"` if `150 <= ms < 400`
- `"slow"` otherwise

Then use a `for` loop to print `classify_latency(ms)` for every value in `test_values` below.

```python
test_values = [50, 200, 400, 999, 149, 150]

def classify_latency(ms):
    if ms < 150:
        return "fast"
    elif 150 <= ms < 400:
        return "moderate"
    else:
        return "slow"

for value in test_values:
    print(f"Latency: {value}ms -> {classify_latency(value)}")

```

## Question 5 (20 pts) — Functions with `*args`, `**kwargs`, and Defaults
Write a function `build_prompt(task, *examples, style="concise", **metadata)` that returns a dict with these keys:
- `"task"` -> the task string
- `"examples"` -> a list of the extra positional args
- `"style"` -> the style (default `"concise"`)
- `"metadata"` -> a dict of any extra keyword arguments

Call it at least twice: once using only `task`, and once supplying extra examples, a custom `style`, and at least one extra keyword argument.

```python
def build_prompt(task, *examples, style="concise", **metadata):
    return {
        "task": task,
        "examples": list(examples),
        "style": style,
        "metadata": metadata
    }

# Call 1: using only task
result_1 = build_prompt("Summarize document")
print("Call 1:", result_1)

# Call 2: extra examples, custom style, and extra keyword arg
result_2 = build_prompt(
    "Translate to French", 
    "Hello -> Bonjour", 
    "Goodbye -> Au revoir", 
    style="formal", 
    author="Landon", 
    version=1.0
)
print("Call 2:", result_2)

```

## Question 6 (20 pts) — JSON
Given the JSON string `raw_response` below, parse it, then print:
1. The value of `"model"`
2. The `"content"` of the **first** message in `"choices"`
3. A re-serialized, pretty-printed (indent=2) version of the whole parsed object

```python
import json
raw_response = '''
{
  "model": "demo-llm-v1",
  "choices": [
    {"role": "assistant", "content": "The capital of Ohio is Columbus."},
    {"role": "assistant", "content": "(alternate) Columbus is the capital."}
  ]
}
'''

parsed_data = json.loads(raw_response)

print("Model:", parsed_data["model"])

print("First message content:", parsed_data["choices"][0]["content"])

pretty_printed = json.dumps(parsed_data, indent=2)
print("\nPretty-printed JSON:\n", pretty_printed)

```
